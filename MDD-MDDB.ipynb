# MDD/MDDB normals over the same interval, showing that the time-dependent boundary orientation agrees with timing and demonstrating the structure's dimensionality

Use MDD on the four MMS FGM measurements to derive a time-dependent local boundary-normal direction, then compare that normal with your timing normal and with the expected nominal magnetopause normal. MDD is built from the full magnetic-field gradient tensor, so it gives you the direction of strongest spatial variation at each instant. For a locally 1-D boundary, that maximum-gradient direction is the boundary normal.

## Helper functions

```python
# ============================================================
# Function to load MMS position from MEC
# ============================================================

def create_mms_position_dataframe(probe, trange):
    """
    Load MMS spacecraft position from MEC and return a
    DataFrame containing GSE position in km.

    Output
    ------
    index : DatetimeIndex
        UTC time

    columns:
        X_gse_km
        Y_gse_km
        Z_gse_km
    """

    # --------------------------------------------------------
    # Load MEC ephemeris data
    # --------------------------------------------------------

    mec_data = pyspedas.projects.mms.mec(
        trange=trange,
        probe=str(probe),
        data_rate="srvy",
        level="l2",
        time_clip=True
    )


    # --------------------------------------------------------
    # MMS MEC GSE position variable
    # --------------------------------------------------------

    varname = f"mms{probe}_mec_r_gse"

    pos = pyspedas.get_data(varname)


    if pos is None:
        raise ValueError(
            f"{varname} was not returned by PySPEDAS.\n"
            f"MEC returned:\n{mec_data}"
        )


    # --------------------------------------------------------
    # Extract data
    # --------------------------------------------------------

    times = np.asarray(pos.times)
    R = np.asarray(pos.y)


    print(
        f"MMS{probe}: "
        f"{varname} shape = {R.shape}"
    )


    # --------------------------------------------------------
    # Convert timestamps
    # --------------------------------------------------------

    datetime = pd.to_datetime(
        times,
        unit="s",
        origin="unix",
        utc=True
    ).tz_convert(None)


    # --------------------------------------------------------
    # Construct dataframe
    #
    # MEC r_gse is normally returned in km.
    # --------------------------------------------------------

    df = pd.DataFrame(
        {
            "X_gse_km": R[:, 0],
            "Y_gse_km": R[:, 1],
            "Z_gse_km": R[:, 2],
        },
        index=datetime
    )

    df.index.name = "time"

    df = (
        df
        .sort_index()
        .loc[
            ~df.index.duplicated(keep="first")
        ]
    )

    return df
```

```python

# =========================================================
# HELPER FUNCTION
# =========================================================

def create_mms_fgm_dataframe(probe, trange):
    """
    Load MMS FGM burst L2 GSE magnetic-field data with
    PySPEDAS and return a standardized pandas DataFrame.

    Output columns:
        Bx
        By
        Bz
        Bmag

    Index:
        UTC DatetimeIndex
    """

    # -----------------------------------------------------
    # Load FGM burst data
    # -----------------------------------------------------

    pyspedas.projects.mms.fgm(
        trange=trange,
        probe=str(probe),
        data_rate="brst",
        level="l2",
        time_clip=True
    )


    # -----------------------------------------------------
    # Retrieve magnetic-field variable
    # -----------------------------------------------------

    varname = f"mms{probe}_fgm_b_gse_brst_l2"

    data = pyspedas.get_data(
        varname
    )


    if data is None:
        raise ValueError(
            f"{varname} was not returned by PySPEDAS."
        )


    # -----------------------------------------------------
    # PySPEDAS output
    #
    # data.times -> UNIX seconds
    # data.y     -> [Bx, By, Bz, |B|]
    # -----------------------------------------------------

    times = np.asarray(
        data.times
    )

    B = np.asarray(
        data.y
    )


    if B.ndim != 2 or B.shape[1] < 3:
        raise ValueError(
            f"Unexpected shape for {varname}: {B.shape}"
        )


    # -----------------------------------------------------
    # Convert time to UTC datetime
    # -----------------------------------------------------

    datetime = pd.to_datetime(
        times,
        unit="s",
        origin="unix",
        utc=True
    ).tz_convert(None)


    # -----------------------------------------------------
    # Build dataframe
    # -----------------------------------------------------

    df = pd.DataFrame(
        {
            "Bx": B[:, 0],
            "By": B[:, 1],
            "Bz": B[:, 2],
        },
        index=datetime
    )


    # -----------------------------------------------------
    # Use supplied |B| if available, otherwise calculate it
    # -----------------------------------------------------

    if B.shape[1] >= 4:

        df["Bmag"] = B[:, 3]

    else:

        df["Bmag"] = np.sqrt(
            df["Bx"]**2
            + df["By"]**2
            + df["Bz"]**2
        )


    # -----------------------------------------------------
    # Clean dataframe
    # -----------------------------------------------------

    df.index.name = "time"

    df = (
        df
        .sort_index()
        .loc[
            ~df.index.duplicated(
                keep="first"
            )
        ]
    )


    # -----------------------------------------------------
    # Keep UNIX time too
    # This can be useful for high-precision timing work.
    # -----------------------------------------------------

    df["unix_time"] = (
        df.index.view("int64")
        / 1e9
    )


    return df
```

```python
# ============================================================
# COMMON TIME GRID
# ============================================================

# Use MMS2 native timestamps as reference
common_times = FGM[2].loc[t0:t1].index


def interpolate_vector_to_times(
    df,
    target_times,
    cols
):
    """
    Interpolate dataframe vector columns onto target_times.
    """

    work = df[list(cols)].copy()

    new_index = (
        work.index
        .union(target_times)
    )

    work = (
        work
        .reindex(new_index)
        .sort_index()
        .interpolate(method="time")
    )

    return (
        work
        .loc[target_times, list(cols)]
        .to_numpy(dtype=float)
    )
```

## Calculate the magnetic gradient tensor
At each time, approximate the magnetic field as locally linear:

$B(r)=B_0+(∇B)⋅(r−r_0)$

With four spacecraft, you can solve the spatial gradients of $B_x,B_y,B_z$ independently.


```python
# ============================================================
# MAGNETIC GRADIENT
# ============================================================

def magnetic_gradient_4sc(
    positions,
    Bvals
):
    """
    Estimate ∇B from four-spacecraft measurements.

    Parameters
    ----------
    positions : (4,3)
        Spacecraft positions in km.

    Bvals : (4,3)
        Magnetic field in nT.

    Returns
    -------
    gradB : (3,3)

    Convention:

        gradB[i,j] = dB_i / dx_j

    Rows:
        Bx, By, Bz

    Columns:
        x, y, z
    """

    # Use spacecraft barycenter
    r0 = np.mean(
        positions,
        axis=0
    )

    B0 = np.mean(
        Bvals,
        axis=0
    )


    dr = positions - r0
    dB = Bvals - B0


    # Least-squares solution:
    #
    # dr @ gradB.T = dB
    #
    gradBT, _, _, _ = np.linalg.lstsq(
        dr,
        dB,
        rcond=None
    )

    gradB = gradBT.T

    return gradB
```

```python
# now calculate it for every timestamp:
gradB_all = np.empty(
    (
        len(common_times),
        3,
        3
    )
)


for k in range(
    len(common_times)
):

    gradB_all[k] = magnetic_gradient_4sc(
        R[k],
        B[k]
    )

    # units are approximately nT/km
```

# Perform MDD
Now form the directional-derivative matrix.

With the convention
$G_{ij} =\frac{∂B_i}{∂x_j}$

the spatial-direction matrix is
$L=G^T G$

```python
# ============================================================
# MDD
# ============================================================

lambda_max = np.zeros(
    len(common_times)
)

lambda_mid = np.zeros(
    len(common_times)
)

lambda_min = np.zeros(
    len(common_times)
)


Nmax = np.zeros(
    (len(common_times), 3)
)

Nmid = np.zeros(
    (len(common_times), 3)
)

Nmin = np.zeros(
    (len(common_times), 3)
)


for k in range(
    len(common_times)
):

    G = gradB_all[k]

    # MDD matrix
    Lmat = G.T @ G

    # Symmetric eigensolver
    vals, vecs = np.linalg.eigh(
        Lmat
    )

    # Sort largest -> smallest
    order = np.argsort(
        vals
    )[::-1]

    vals = vals[order]
    vecs = vecs[:, order]


    lambda_max[k] = vals[0]
    lambda_mid[k] = vals[1]
    lambda_min[k] = vals[2]


    Nmax[k] = vecs[:, 0]
    Nmid[k] = vecs[:, 1]
    Nmin[k] = vecs[:, 2]
```

 The actual directional-gradient amplitudes plotted by Shi et al. are often

$\sqrt{λ_{max}},\sqrt{λ_{mid}},\sqrt{λ_{min}}$



```python
sqrt_lmax = np.sqrt(
    np.maximum(lambda_max, 0)
)

sqrt_lmid = np.sqrt(
    np.maximum(lambda_mid, 0)
)

sqrt_lmin = np.sqrt(
    np.maximum(lambda_min, 0)
)
```

## Resolve the MDD sign ambiguity

MDD eigenvectors still have a ± ambiguity.

This is where your timing result becomes extremely useful. Use the $N_{timing}$ obtained from four_spacecraft_timing_normals.ipynb



Then orient every MDD normal toward the timing normal:

```python
def orient_vectors(
    vectors,
    reference
):

    reference = (
        reference
        / np.linalg.norm(reference)
    )

    out = vectors.copy()

    for i in range(len(out)):

        if np.dot(
            out[i],
            reference
        ) < 0:

            out[i] *= -1

    return out
```

## Calculate dimensionality indices

Shi et al. discuss the Rezeau-type indices:

$D_1 = \frac{\sqrt{λ_{max}} -\sqrt{λ_{mid}}}{\sqrt{λ_{max}}},$

$D_2 = \frac{\sqrt{λ_{mid}} -\sqrt{λ_{min}}}{\sqrt{λ_{max}}},$

$D_3 = \frac{\sqrt{λ_{min}}}{\sqrt{λ_{max}}}$


These help tell you whether the structure is approximately 1-D, 2-D, or 3-D

```python
eps = 1e-30

D1 = (
    sqrt_lmax - sqrt_lmid
) / (
    sqrt_lmax + eps
)

D2 = (
    sqrt_lmid - sqrt_lmin
) / (
    sqrt_lmax + eps
)

D3 = (
    sqrt_lmin
) / (
    sqrt_lmax + eps
)
```

nterpret approximately:

    - $D_1$ large → predominantly 1-D.
    - $D_2$ large → significant 2-D character.
    - $D_3$ large → significant 3-D variation.

For your KH boundary-normal argument, you especially want times when

$λ_{max} ≫λ_{mid},λ_{min}$

	​
Then $N_{max}$ is a defensible local boundary normal.

# Add a $∇⋅B$ quality check

Shi et al. emphasize checking errors in the magnetic-gradient estimate. A useful diagnostic is

$Q= \frac{|∇⋅B|}{max∣∂B_i /∂x_j|}$


```python
divB = np.zeros(
    len(common_times)
)

quality = np.zeros(
    len(common_times)
)


for k in range(
    len(common_times)
):

    G = gradB_all[k]

    divB[k] = np.trace(
        G
    )

    max_grad = np.max(
        np.abs(G)
    )

    if max_grad > 0:

        quality[k] = (
            np.abs(divB[k])
            / max_grad
        )

    else:

        quality[k] = np.nan

```

Shi et al. discuss values around 0.4 or larger as a warning that the gradient/current estimate may be unreliable.

# What you're looking for:

E.g., For Event 4, during the actual boundary transition around ~22:18:20–22:18:21, I would look for four things happening together:

- $\sqrt{λ_{max}}≫ \sqrt{λ_{mid}}, \sqrt{λ_{min}}$

- a large $D_1$: $D_1$ →1,

- a relatively stable $N_{MDD}(t)$

- and a low gradient-error indicator.


If those conditions hold, you can say that the magnetic structure is locally quasi-1-D and that the MDD maximum-gradient direction provides an independent estimate of its normal. Because MDD works point-by-point, it can also reveal time-dependent normal rotations that an interval-based MVA cannot.

```python
def angle_between_vectors(a, b):

    a = np.asarray(a)
    b = np.asarray(b)

    a = a / np.linalg.norm(a)
    b = b / np.linalg.norm(b)

    return np.degrees(
        np.arccos(
            np.clip(
                np.dot(a, b),
                -1,
                1
            )
        )
    )
```

```python
import numpy as np
import pandas as pd



# ============================================================
# SELECT EVENT  CORE
# ============================================================

mask = (
    (common_times >= t0)
    & (common_times <= t1)
)


# ============================================================
# 1. EIGENVALUE SEPARATION
# ============================================================

mean_lmax = np.nanmean(
    sqrt_lmax[mask]
)

mean_lmid = np.nanmean(
    sqrt_lmid[mask]
)

mean_lmin = np.nanmean(
    sqrt_lmin[mask]
)


ratio_max_mid = (
    mean_lmax / mean_lmid
)

ratio_mid_min = (
    mean_lmid / mean_lmin
)


# ============================================================
# 2. DIMENSIONALITY INDEX D1
# ============================================================

mean_D1 = np.nanmean(
    D1[mask]
)

median_D1 = np.nanmedian(
    D1[mask]
)

fraction_D1_good = np.mean(
    D1[mask] > 0.5
)


# ============================================================
# 3. MDD NORMAL STABILITY
# ============================================================

N_event = Nmax_oriented[mask]


# Mean normal
N_mdd_mean = np.nanmean(
    N_event,
    axis=0
)

N_mdd_mean = (
    N_mdd_mean
    / np.linalg.norm(N_mdd_mean)
)


def angle_between(a, b):

    a = np.asarray(a, dtype=float)
    b = np.asarray(b, dtype=float)

    a = a / np.linalg.norm(a)
    b = b / np.linalg.norm(b)

    return np.degrees(
        np.arccos(
            np.clip(
                np.dot(a, b),
                -1,
                1
            )
        )
    )


# Angular deviation of every MDD normal
# from the mean MDD normal
normal_deviation = np.array([
    angle_between(
        n,
        N_mdd_mean
    )
    for n in N_event
])


mean_normal_deviation = np.nanmean(
    normal_deviation
)

median_normal_deviation = np.nanmedian(
    normal_deviation
)

max_normal_deviation = np.nanmax(
    normal_deviation
)


# Compare mean MDD normal to timing normal
mdd_timing_angle = angle_between(
    N_mdd_mean,
    N_timing
)


# ============================================================
# 4. GRADIENT QUALITY
# ============================================================

mean_quality = np.nanmean(
    quality[mask]
)

median_quality = np.nanmedian(
    quality[mask]
)

fraction_quality_good = np.mean(
    quality[mask] < 0.4
)


# ============================================================
# PRINT RESULTS
# ============================================================

print("=" * 65)
print(f"{event_num} MDD DIAGNOSTICS")
print(
    f"Interval: {t0.time()} - "
    f"{t1.time()} UT"
)
print("=" * 65)


print("\n1. MDD EIGENVALUE SEPARATION")
print("-" * 40)

print(
    f"<sqrt(lambda_max)> = "
    f"{mean_lmax:.5f} nT/km"
)

print(
    f"<sqrt(lambda_mid)> = "
    f"{mean_lmid:.5f} nT/km"
)

print(
    f"<sqrt(lambda_min)> = "
    f"{mean_lmin:.5f} nT/km"
)

print(
    f"lambda_max^0.5 / lambda_mid^0.5 = "
    f"{ratio_max_mid:.2f}"
)

print(
    f"lambda_mid^0.5 / lambda_min^0.5 = "
    f"{ratio_mid_min:.2f}"
)


print("\n2. ONE-DIMENSIONALITY D1")
print("-" * 40)

print(
    f"Mean D1   = {mean_D1:.3f}"
)

print(
    f"Median D1 = {median_D1:.3f}"
)

print(
    f"Fraction of samples with D1 > 0.5 = "
    f"{fraction_D1_good*100:.1f}%"
)


print("\n3. MDD NORMAL STABILITY")
print("-" * 40)

print(
    "Mean MDD normal [GSE] = "
    f"[{N_mdd_mean[0]:+.3f}, "
    f"{N_mdd_mean[1]:+.3f}, "
    f"{N_mdd_mean[2]:+.3f}]"
)

print(
    f"Mean angular deviation from "
    f"<N_MDD> = "
    f"{mean_normal_deviation:.1f}°"
)

print(
    f"Median angular deviation = "
    f"{median_normal_deviation:.1f}°"
)

print(
    f"Maximum angular deviation = "
    f"{max_normal_deviation:.1f}°"
)

print(
    f"Angle between <N_MDD> and "
    f"N_timing = "
    f"{mdd_timing_angle:.1f}°"
)


print("\n4. GRADIENT QUALITY")
print("-" * 40)

print(
    f"Mean quality metric = "
    f"{mean_quality:.3f}"
)

print(
    f"Median quality metric = "
    f"{median_quality:.3f}"
)

print(
    f"Fraction with quality < 0.4 = "
    f"{fraction_quality_good*100:.1f}%"
)


print("\n" + "=" * 65)




print(
    "Mean MDD normal:",
    N_mdd_mean
)

print(
    "Timing normal:",
    N_timing
)

print(
    "Cross-product normal:",
    N_cross
)
```

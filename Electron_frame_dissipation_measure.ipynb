# Align dataframes

```python
import numpy as np
import pandas as pd


# ============================================================
# TIME-ALIGN MMS DATAFRAMES TO A COMMON TIME GRID
# ============================================================

def align_to_reference(reference_df, df, max_gap=None):
    """
    Interpolate a DataFrame onto the timestamps of a reference
    DataFrame.

    Parameters
    ----------
    reference_df : pd.DataFrame
        DataFrame whose DatetimeIndex defines the output time grid.

    df : pd.DataFrame
        DataFrame to interpolate onto the reference time grid.

    max_gap : str or pd.Timedelta, optional
        Maximum allowed distance between an output timestamp and
        the nearest original measurement. Examples:
            "100ms"
            "200ms"
            "1s"

        Points farther away are replaced with NaN.

    Returns
    -------
    aligned : pd.DataFrame
        DataFrame interpolated onto reference_df.index.
    """

    # --------------------------------------------------------
    # Clean reference index
    # --------------------------------------------------------

    ref = reference_df.copy()

    ref.index = pd.to_datetime(ref.index)

    ref = (
        ref[~ref.index.duplicated(keep="first")]
        .sort_index()
    )


    # --------------------------------------------------------
    # Clean source index
    # --------------------------------------------------------

    source = df.copy()

    source.index = pd.to_datetime(source.index)

    source = (
        source[~source.index.duplicated(keep="first")]
        .sort_index()
    )


    # --------------------------------------------------------
    # Create combined time grid
    #
    # This is important:
    # interpolate using ORIGINAL source timestamps first,
    # then select reference timestamps.
    # --------------------------------------------------------

    combined_index = (
        source.index
        .union(ref.index)
        .sort_values()
    )


    temp = source.reindex(combined_index)


    # --------------------------------------------------------
    # Time interpolation
    # --------------------------------------------------------

    temp = temp.interpolate(
        method="time",
        limit_area="inside"
    )


    # --------------------------------------------------------
    # Return only reference timestamps
    # --------------------------------------------------------

    aligned = temp.reindex(ref.index)


    # --------------------------------------------------------
    # OPTIONAL:
    # Reject points too far from an actual source measurement
    # --------------------------------------------------------

    if max_gap is not None:

        max_gap = pd.Timedelta(max_gap)

        source_times = source.index.values
        ref_times = ref.index.values

        idx = np.searchsorted(
            source_times,
            ref_times
        )

        idx_before = np.clip(
            idx - 1,
            0,
            len(source_times) - 1
        )

        idx_after = np.clip(
            idx,
            0,
            len(source_times) - 1
        )

        before = source_times[idx_before]
        after = source_times[idx_after]

        dt_before = np.abs(
            ref_times - before
        )

        dt_after = np.abs(
            after - ref_times
        )

        nearest_dt = np.minimum(
            dt_before,
            dt_after
        )

        valid = nearest_dt <= max_gap.to_timedelta64()

        aligned.loc[~valid, :] = np.nan


    return aligned
```

```python
# ============================================================
# USE DES ELECTRON TIMESTAMPS AS MASTER GRID
# ============================================================

Ve_aligned = (
    fpi_e_lmn
    .copy()
)

Ve_aligned.index = pd.to_datetime(
    Ve_aligned.index
)

Ve_aligned = (
    Ve_aligned[
        ~Ve_aligned.index.duplicated(keep="first")
    ]
    .sort_index()
)
```

```python
# Magnetic field
B_aligned = align_to_reference(
    Ve_aligned,
    fgm_lmn,
    max_gap="100ms"
)


# Electric field
E_aligned = align_to_reference(
    Ve_aligned,
    edp_lmn,
    max_gap="100ms"
)


# Ion moments
#
# DIS has lower cadence, so allow a somewhat
# larger nearest-measurement distance.
Vi_aligned = align_to_reference(
    Ve_aligned,
    fpi_i_lmn,
    max_gap="200ms"
)
```

calculate the electron-frame dissipation measure

$$ D_e \approx \mathbf J\cdot\mathbf E', \qquad \mathbf E'=\mathbf E+\mathbf V_e\times\mathbf B, $$

with

$$ \mathbf J \approx e n_e(\mathbf V_i-\mathbf V_e) $$

assuming quasi-neutrality.

The biggest thing to get right is the units: \(B\) nT → T, \(E\) mV/m → V/m, \(V\) km/s → m/s, and \(n_e\) cm\(^{-3}\) → m\(^{-3}\).

```python
# ============================================================
# STEP 1: EXTRACT TIME-ALIGNED LMN VARIABLES
# ============================================================

import numpy as np
import pandas as pd


# Magnetic field [nT]
B = master[
    [
        "B_BL",
        "B_BM",
        "B_BN"
    ]
].to_numpy(dtype=float)


# Electric field [mV/m]
E = master[
    [
        "E_EL",
        "E_EM",
        "E_EN"
    ]
].to_numpy(dtype=float)


# Ion bulk velocity [km/s]
Vi = master[
    [
        "Vi_ViL",
        "Vi_ViM",
        "Vi_ViN"
    ]
].to_numpy(dtype=float)


# Electron bulk velocity [km/s]
Ve = master[
    [
        "Ve_VeL",
        "Ve_VeM",
        "Ve_VeN"
    ]
].to_numpy(dtype=float)


# Electron number density [cm^-3]
ne = master[
    "Ve_ne"
].to_numpy(dtype=float)


print("B shape :", B.shape)
print("E shape :", E.shape)
print("Vi shape:", Vi.shape)
print("Ve shape:", Ve.shape)
print("ne shape:", ne.shape)
```

```python
# ============================================================
# STEP 2: CONVERT TO SI UNITS
# ============================================================

# Elementary charge [C]
qe = 1.602176634e-19


# B: nT -> T
B_SI = B * 1e-9


# E: mV/m -> V/m
E_SI = E * 1e-3


# Velocity: km/s -> m/s
Vi_SI = Vi * 1e3
Ve_SI = Ve * 1e3


# Density: cm^-3 -> m^-3
ne_SI = ne * 1e6


print("B units: T")
print("E units: V/m")
print("Vi, Ve units: m/s")
print("ne units: m^-3")
```

```python
# ============================================================
# STEP 3: CALCULATE CURRENT DENSITY
#
# J = e ne (Vi - Ve)
# ============================================================

J = (
    qe
    * ne_SI[:, None]
    * (Vi_SI - Ve_SI)
)


# LMN components
JL = J[:, 0]
JM = J[:, 1]
JN = J[:, 2]


print("J shape:", J.shape)

print(
    "Current density range:",
    np.nanmin(J * 1e9),
    "to",
    np.nanmax(J * 1e9),
    "nA/m^2"
)
```

```python
# ============================================================
# STEP 4: CALCULATE Ve x B
# ============================================================

Ve_cross_B = np.cross(
    Ve_SI,
    B_SI
)


print(
    "Ve x B shape:",
    Ve_cross_B.shape
)
```

```python
# ============================================================
# STEP 5: ELECTRON-FRAME NONIDEAL ELECTRIC FIELD
#
# E' = E + Ve x B
# ============================================================

Eprime = (
    E_SI
    + Ve_cross_B
)


# LMN components
Eprime_L = Eprime[:, 0]
Eprime_M = Eprime[:, 1]
Eprime_N = Eprime[:, 2]


print(
    "E' range:",
    np.nanmin(Eprime * 1e3),
    "to",
    np.nanmax(Eprime * 1e3),
    "mV/m"
)
```

Step 6: Calculate \(D_e = \mathbf J\cdot\mathbf E'\)

Now take the dot product:

$$ D_e = J_LE'_L + J_ME'_M + J_NE'_N. $$

```python
# ============================================================
# STEP 6: ELECTRON-FRAME DISSIPATION
#
# De = J · E'
# ============================================================

De = np.sum(
    J * Eprime,
    axis=1
)


print(
    "De range:",
    np.nanmin(De * 1e9),
    "to",
    np.nanmax(De * 1e9),
    "nW/m^3"
)
```

```python
# ============================================================
# STEP 7: LMN CONTRIBUTIONS TO De
# ============================================================

De_L = (
    JL
    * Eprime_L
)

De_M = (
    JM
    * Eprime_M
)

De_N = (
    JN
    * Eprime_N
)
```

```python
# asnswer should be 0

De_components = (
    De_L
    + De_M
    + De_N
)


print(
    "Maximum difference between "
    "De and component sum:",
    np.nanmax(
        np.abs(
            De - De_components
        )
    )
)
```

```python
# ============================================================
# STEP 8: CREATE De DATAFRAME
# ============================================================

De_df = pd.DataFrame(
    index=master.index
)


# ------------------------------------------------------------
# Current density [nA/m^2]
# ------------------------------------------------------------

De_df["JL"] = JL * 1e9
De_df["JM"] = JM * 1e9
De_df["JN"] = JN * 1e9


# ------------------------------------------------------------
# Electron-frame nonideal electric field [mV/m]
# ------------------------------------------------------------

De_df["Eprime_L"] = (
    Eprime_L * 1e3
)

De_df["Eprime_M"] = (
    Eprime_M * 1e3
)

De_df["Eprime_N"] = (
    Eprime_N * 1e3
)


# ------------------------------------------------------------
# Individual dissipation components [nW/m^3]
# ------------------------------------------------------------

De_df["De_L"] = (
    De_L * 1e9
)

De_df["De_M"] = (
    De_M * 1e9
)

De_df["De_N"] = (
    De_N * 1e9
)


# ------------------------------------------------------------
# Total electron-frame dissipation [nW/m^3]
# ------------------------------------------------------------

De_df["De"] = (
    De * 1e9
)


De_df.index.name = "datetime"


display(De_df.head())
```

```python
# ============================================================
# STEP 9: QUALITY CHECK
# ============================================================

print("NaN counts:")
print(De_df.isna().sum())

print()


print("Current density [nA/m^2]:")
print(
    De_df[
        ["JL", "JM", "JN"]
    ].describe()
)

print()


print("E' [mV/m]:")
print(
    De_df[
        [
            "Eprime_L",
            "Eprime_M",
            "Eprime_N"
        ]
    ].describe()
)

print()


print("De [nW/m^3]:")
print(
    De_df[
        [
            "De_L",
            "De_M",
            "De_N",
            "De"
        ]
    ].describe()
)
```

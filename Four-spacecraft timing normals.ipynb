# Helper Functions

```python
# ============================================================
# Function to load MMS position from MEC
# ============================================================

def create_mms_position_dataframe(probe, trange):
    """
    Load MMS spacecraft position from MEC and return a
    DataFrame containing GSE position in km.

    Output
    ------
    index : DatetimeIndex
        UTC time

    columns:
        X_gse_km
        Y_gse_km
        Z_gse_km
    """

    # --------------------------------------------------------
    # Load MEC ephemeris data
    # --------------------------------------------------------

    mec_data = pyspedas.projects.mms.mec(
        trange=trange,
        probe=str(probe),
        data_rate="srvy",
        level="l2",
        time_clip=True
    )


    # --------------------------------------------------------
    # MMS MEC GSE position variable
    # --------------------------------------------------------

    varname = f"mms{probe}_mec_r_gse"

    pos = pyspedas.get_data(varname)


    if pos is None:
        raise ValueError(
            f"{varname} was not returned by PySPEDAS.\n"
            f"MEC returned:\n{mec_data}"
        )


    # --------------------------------------------------------
    # Extract data
    # --------------------------------------------------------

    times = np.asarray(pos.times)
    R = np.asarray(pos.y)


    print(
        f"MMS{probe}: "
        f"{varname} shape = {R.shape}"
    )


    # --------------------------------------------------------
    # Convert timestamps
    # --------------------------------------------------------

    datetime = pd.to_datetime(
        times,
        unit="s",
        origin="unix",
        utc=True
    ).tz_convert(None)


    # --------------------------------------------------------
    # Construct dataframe
    #
    # MEC r_gse is normally returned in km.
    # --------------------------------------------------------

    df = pd.DataFrame(
        {
            "X_gse_km": R[:, 0],
            "Y_gse_km": R[:, 1],
            "Z_gse_km": R[:, 2],
        },
        index=datetime
    )

    df.index.name = "time"

    df = (
        df
        .sort_index()
        .loc[
            ~df.index.duplicated(keep="first")
        ]
    )

    return df
```

```python
# =========================================================
# HELPER FUNCTION
# =========================================================

def create_mms_fgm_dataframe(probe, trange):
    """
    Load MMS FGM burst L2 GSE magnetic-field data with
    PySPEDAS and return a standardized pandas DataFrame.

    Output columns:
        Bx
        By
        Bz
        Bmag

    Index:
        UTC DatetimeIndex
    """

    # -----------------------------------------------------
    # Load FGM burst data
    # -----------------------------------------------------

    pyspedas.projects.mms.fgm(
        trange=trange,
        probe=str(probe),
        data_rate="brst",
        level="l2",
        time_clip=True
    )


    # -----------------------------------------------------
    # Retrieve magnetic-field variable
    # -----------------------------------------------------

    varname = f"mms{probe}_fgm_b_gse_brst_l2"

    data = pyspedas.get_data(
        varname
    )


    if data is None:
        raise ValueError(
            f"{varname} was not returned by PySPEDAS."
        )


    # -----------------------------------------------------
    # PySPEDAS output
    #
    # data.times -> UNIX seconds
    # data.y     -> [Bx, By, Bz, |B|]
    # -----------------------------------------------------

    times = np.asarray(
        data.times
    )

    B = np.asarray(
        data.y
    )


    if B.ndim != 2 or B.shape[1] < 3:
        raise ValueError(
            f"Unexpected shape for {varname}: {B.shape}"
        )


    # -----------------------------------------------------
    # Convert time to UTC datetime
    # -----------------------------------------------------

    datetime = pd.to_datetime(
        times,
        unit="s",
        origin="unix",
        utc=True
    ).tz_convert(None)


    # -----------------------------------------------------
    # Build dataframe
    # -----------------------------------------------------

    df = pd.DataFrame(
        {
            "Bx": B[:, 0],
            "By": B[:, 1],
            "Bz": B[:, 2],
        },
        index=datetime
    )


    # -----------------------------------------------------
    # Use supplied |B| if available, otherwise calculate it
    # -----------------------------------------------------

    if B.shape[1] >= 4:

        df["Bmag"] = B[:, 3]

    else:

        df["Bmag"] = np.sqrt(
            df["Bx"]**2
            + df["By"]**2
            + df["Bz"]**2
        )


    # -----------------------------------------------------
    # Clean dataframe
    # -----------------------------------------------------

    df.index.name = "time"

    df = (
        df
        .sort_index()
        .loc[
            ~df.index.duplicated(
                keep="first"
            )
        ]
    )


    # -----------------------------------------------------
    # Keep UNIX time too
    # This can be useful for high-precision timing work.
    # -----------------------------------------------------

    df["unix_time"] = (
        df.index.view("int64")
        / 1e9
    )


    return df
```

```python
def find_max_gradient_time(
    df,
    column,
    start,
    end,
    smooth_ms=100
):
    """
    Find the time of maximum absolute temporal gradient
    in a magnetic-field component.
    """

    x = df.loc[
        start:end,
        column
    ].dropna().copy()

    if len(x) < 5:
        return pd.NaT

    # Optional short smoothing
    if smooth_ms is not None:

        x_smooth = (
            x.rolling(
                f"{smooth_ms}ms",
                center=True,
                min_periods=1
            )
            .mean()
        )

    else:
        x_smooth = x


    # Time in seconds
    tsec = (
        x_smooth.index.view("int64")
        / 1e9
    )

    y = x_smooth.to_numpy()


    # dB/dt
    dydt = np.gradient(
        y,
        tsec
    )


    i = np.nanargmax(
        np.abs(dydt)
    )


    return x_smooth.index[i]
```

```python
# interpolate each trajectory
def interpolate_position(df, target_time):
    """
    Linearly interpolate spacecraft position to target_time.

    Returns
    -------
    numpy array
        [X, Y, Z] in km
    """

    # Time relative to first sample
    t0 = df.index[0]

    t = (
        df.index - t0
    ).total_seconds().to_numpy()

    target = (
        target_time - t0
    ).total_seconds()


    r = np.array([
        np.interp(
            target,
            t,
            df[col].to_numpy()
        )
        for col in [
            "X_gse_km",
            "Y_gse_km",
            "Z_gse_km"
        ]
    ])

    return r
```

```python
def four_spacecraft_timing(
    crossing_times,
    positions,
    reference_sc=1
):
    """
    Four-spacecraft timing solution.

    Solves:

        dt_i = dr_i dot m

    where

        m = n / Vn

    Returns
    -------
    N : unit boundary normal
    Vn : normal propagation speed [km/s]
    m : timing vector [s/km]
    """

    tref = crossing_times[
        reference_sc
    ]

    rref = np.asarray(
        positions[reference_sc],
        dtype=float
    )


    A = []
    b = []


    for sc in sorted(
        crossing_times.keys()
    ):

        if sc == reference_sc:
            continue


        ri = np.asarray(
            positions[sc],
            dtype=float
        )


        dr = ri - rref


        dt = (
            crossing_times[sc]
            - tref
        ).total_seconds()


        A.append(dr)

        b.append(dt)


    A = np.asarray(A)
    b = np.asarray(b)


    # Solve:
    #
    # A @ m = b
    #
    m = np.linalg.solve(
        A,
        b
    )


    m_norm = np.linalg.norm(m)


    Vn = 1.0 / m_norm

    N = m / m_norm


    return N, Vn, m
```

# Step 7: compare it with your cross-product normal

The best cross-product normal for the event has a ± ambiguity, therefore compare both signs:

```python
N_cross = (
    N_cross
    / np.linalg.norm(N_cross)
)


def angle_deg(a, b):

    a = a / np.linalg.norm(a)
    b = b / np.linalg.norm(b)

    c = np.clip(
        np.dot(a, b),
        -1,
        1
    )

    return np.degrees(
        np.arccos(c)
    )


angle_plus = angle_deg(
    N_timing,
    N_cross
)

angle_minus = angle_deg(
    N_timing,
    -N_cross
)


print(f"N component comparison for {event_num}")
print('Ncross',N_cross)
print('Ntiming',N_timing)
print(
    f"+Ncross difference = "
    f"{angle_plus:.1f}°"
)

print(
    f"-Ncross difference = "
    f"{angle_minus:.1f}°"
)
```

That immediately tells you which sign of your cross-product normal was physically correct.

If one of the Ncross differences has a lower value, that one is physically correct.

Since the timing analysis resolves the propagation direction, the cross-product normal should be oriented to agree with that direction. 


The a value <= 20 degrees, the agreement means the independently determined timing normal and your cross-product normal are nearly parallel.

# Step 8: test the planar-boundary assumption

This matters for the paper.

Because four spacecraft give exactly enough information to solve the three unknown components of m, a single four-spacecraft timing solution has no redundant measurement to tell you whether the fit is actually good.

So I would repeat the timing using several recognizable positions across the same boundary.

For example, define crossings by:

- maximum $∣dB_y /dt∣$;
- $B_y=0$ or another common level;
- midpoint of the full $B_y$ rotation;
- a density threshold;
- perhaps maximum ∣dB/dt∣.

If those give normals such as

| method   |  Nx | Ny | Nz | Vn |

and they agree within ~10–20°, the timing normal is much more convincing.

Do not cherry-pick the marker producing the most desirable normal.
